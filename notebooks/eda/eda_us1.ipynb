```python
# ============================================================
# US-1 EDA: Dataset Loading, Structure, Quality Checks
# Project: Impact of Climate Change on Energy Demand
# ============================================================

import pandas as pd
from pathlib import Path

# ------------------------------------------------------------
# Paths
# ------------------------------------------------------------

DATA_PATH = Path("../../data/processed/final_merged.csv")
OUTPUT_DIR = Path("../../reports/us1_eda_outputs")

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

# ------------------------------------------------------------
# Load dataset
# ------------------------------------------------------------

df = pd.read_csv(DATA_PATH)

print("Dataset loaded successfully.")
print("Shape:", df.shape)

df.head()
```

```python
# ============================================================
# Dataset Structure
# ============================================================

print("Dataset Columns:")

for col in df.columns:
    print("-", col)

print("\nData Types:")
print(df.dtypes)

print("\nDataset Info:")
df.info()
```

```python
# Summary Statistics 
print("\nSummary Statistics:")
print(df.describe())
```

```python
# Identify Key Variables    
target_variable = "electricity_demand_per_capita"

key_features = [
    "temperature_change_c",
    "co2_per_capita",
    "gdp",
    "population",
    "renewables_share_elec",
    "fossil_share_elec"
]

print("Target Variable:", target_variable)
print("Key Features:", key_features)
```

```python
# ============================================================
# Missing Value Check
# ============================================================

missing_values = df.isnull().sum()
missing_percent = (df.isnull().mean() * 100).round(2)

missing_summary = pd.DataFrame({
    "missing_count": missing_values,
    "missing_percent": missing_percent
})

print("Missing Value Summary:")
print(missing_summary)

missing_summary.to_csv(OUTPUT_DIR / "missing_values_summary.csv")
```

```python
# Check duplicate records based on country-year

duplicate_count = df.duplicated(subset=["country", "year"]).sum()

print("Duplicate country-year records:", duplicate_count)

duplicate_rows = df[df.duplicated(subset=["country", "year"], keep=False)]

print("\nDuplicate Rows:")
display(duplicate_rows)
```

```python
# ============================================================
# Negative Value Check
# ============================================================

negative_check_cols = [
    "gdp",
    "population",
    "electricity_demand_per_capita",
    "co2_per_capita",
    "renewables_share_elec",
    "fossil_share_elec"
]

negative_results = {}

for col in negative_check_cols:
    if col in df.columns:
        negative_results[col] = int((df[col] < 0).sum())

negative_summary = pd.DataFrame(
    negative_results.items(),
    columns=["column", "negative_value_count"]
)

print("Negative Value Check:")
print(negative_summary)

negative_summary.to_csv(OUTPUT_DIR / "negative_value_check.csv", index=False)
```

```python
# ============================================================
# Key Variable Range Check
# ============================================================

range_check_cols = [
    "temperature_change_c",
    "gdp",
    "population",
    "electricity_demand_per_capita",
    "co2_per_capita",
    "renewables_share_elec",
    "fossil_share_elec"
]

range_results = []

for col in range_check_cols:
    if col in df.columns:
        range_results.append({
            "column": col,
            "min": df[col].min(),
            "max": df[col].max(),
            "mean": df[col].mean(),
            "median": df[col].median()
        })

range_summary = pd.DataFrame(range_results)

print("Range Check Summary:")
print(range_summary)

range_summary.to_csv(OUTPUT_DIR / "range_check_summary.csv", index=False)
```

```python
# Validate country and year consistency

print("Country missing values:", df["country"].isnull().sum())
print("Year missing values:", df["year"].isnull().sum())

print("\nYear data type:", df["year"].dtype)
print("Minimum year:", df["year"].min())
print("Maximum year:", df["year"].max())

invalid_years = df[(df["year"] < 1990) | (df["year"] > 2024)]
print("\nInvalid year records:", len(invalid_years))

blank_countries = df[df["country"].astype(str).str.strip() == ""]
print("Blank country records:", len(blank_countries))

unique_countries = df["country"].nunique()
print("Unique countries:", unique_countries)
```

```python
# ============================================================
# Final US-1 EDA Validation Summary
# ============================================================

validation_summary = {
    "total_rows": df.shape[0],
    "total_columns": df.shape[1],
    "unique_countries": df["country"].nunique(),
    "minimum_year": df["year"].min(),
    "maximum_year": df["year"].max(),
    "duplicate_country_year_rows": int(duplicate_count),
    "total_missing_values": int(df.isnull().sum().sum())
}

validation_summary_df = pd.DataFrame(
    validation_summary.items(),
    columns=["check", "value"]
)

print("Final US-1 EDA Validation Summary:")
print(validation_summary_df)

validation_summary_df.to_csv(
    OUTPUT_DIR / "us1_validation_summary.csv",
    index=False
)
```
